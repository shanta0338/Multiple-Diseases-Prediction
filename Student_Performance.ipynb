```python
from sklearn.preprocessing import LabelEncoder, StandardScaler, MinMaxScaler
from sklearn.model_selection import train_test_split, RandomizedSearchCV
from sklearn.pipeline import Pipeline
```

```python
from sklearn.neural_network import MLPRegressor
from sklearn.linear_model import ElasticNet
from sklearn.neighbors import KNeighborsRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import AdaBoostRegressor, BaggingRegressor, RandomForestRegressor, StackingRegressor
```

```python
from sklearn.ensemble import GradientBoostingRegressor
```

```python
df.columns
```

```python
encoder = LabelEncoder()
for features in ['Parental_Involvement', 'Access_to_Resources', 'Extracurricular_Activities', 'Motivation_Level', 'Internet_Access', 'Family_Income', 'School_Type', 'Peer_Influence', 'Teacher_Quality','Learning_Disabilities', 'Parental_Education_Level', 'Distance_from_Home','Gender']:
    df[features] = encoder.fit_transform(df[features])
```

```python
df.head(10)
```

```python
Y = df.iloc[:, -1]
X = df.iloc[:, 0:-1]
```

```python
x_train, x_test, y_train, y_test = train_test_split(X, Y, test_size=0.2, train_size=0.8, random_state= 42, shuffle= True)
```

```python
base_model = [('KNN', KNeighborsRegressor()), ('Net', ElasticNet())]
```

```python
stack = StackingRegressor(estimators= base_model, final_estimator= GradientBoostingRegressor())
```

```python
pipe = Pipeline([('Scaler', MinMaxScaler()), ('model', stack)])
```

```python
param = {
        'model__KNN__n_neighbors': [5, 10, 15, 20], 
        'model__KNN__weights':['uniform', 'distance'],
        'model__KNN__p':[1, 2],
        'model__KNN__algorithm' : ['auto', 'ball_tree', 'kd_tree', 'brute'],

        'model__final_estimator__loss':['squared_error', 'absolute_error'],
        'model__final_estimator__learning_rate': [0.1, 0.01, 0.001, 0.0001, 0.00001],
        'model__final_estimator__n_estimators': [50, 100, 150, 200]
}
```

```python
model = RandomizedSearchCV(estimator=pipe, param_distributions=param, cv=10, n_iter= 10, n_jobs=-1, verbose=2,)
```

```python
model.fit(x_train, y_train)
```

```python
model.score(x_test, y_test)
```

```python
from sklearn.metrics import r2_score
```

```python
y_pred = model.predict(x_test)
```

```python
score = r2_score(y_test, y_pred)
```
